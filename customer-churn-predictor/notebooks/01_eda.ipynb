```python
import matplotlib.pyplot as plt
import seaborn as sns
from src.load_data import load_churn_csv, basic_clean
```

```python
df = basic_clean(load_churn_csv())

# Always check the shape, dtypes, and missing values before modeling
print(df.shape)
print(df.info())
print(df.isnull().sum())
```

```python
# Churn rate drives metric choice 
churn_rate = df['Churn'].mean()
print(f"Churn rate: {churn_rate:.1%}")
```

```python
# Compare churn vs stay for the numeric columns
for col in ['tenure', 'MonthlyCharges', 'TotalCharges']:
    sns.boxplot(data=df, x='Churn', y=col)
    plt.title(f"{col} by Churn")
    plt.show()
    plt.savefig(f'/Users/bharathkbk.12/ai-ml-journey/customer-churn-predictor/data/{col}_plot.png')
```

```python
# Categorial breakdown 
sns.countplot(data=df, x='Contract', hue='Churn')
plt.show()
plt.savefig('/Users/bharathkbk.12/ai-ml-journey/customer-churn-predictor/data/Churn_by_Contract.png')
```
