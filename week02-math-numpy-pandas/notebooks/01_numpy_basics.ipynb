```python
import numpy as np

# creating arrays
a = np.array([1, 2, 3])                     # 1D vector
b = np.array([[1, 2, 3], [4, 5, 6]])        # 2D matrix (2x3)
zeros = np.zeros((3, 4))                    # 3x4 matrix of zeros
ones = np.ones((2, 3))
random = np.random.randn(3, 3)              # standard normal
identity = np.eye(3)                        # 3x3 identity matrix
```

```python
# shape and dtype
print(a.shape)
print(b.shape)
print(a.dtype)
```

```python
# Indexing and Slicing
print(b[0, 1])          # 2 (row 0, col 1)
print(b[:, 1])          # [2, 5] (all rows, col 1)
print(b[0, :])          # [1, 2, 3] (row 0, all cols)
```

```python
# Reshaping
flat = np.arange(12)            # [0,1,2,...,11]
matrix = flat.reshape(3, 4)     # 3x4 matrix
back = matrix.flatten()         # back to 1D
```

```python
# Basic operations (element-wise)
x = np.array([1, 2, 3])
y = np.array([4, 5, 6])
print(x + y)       
print(x * 2)       
print(x ** 2)      
```

```python
# Aggregation
data = np.array([1, 2, 3, 4, 5])
print(data.sum())
print(data.mean())
print(data.std())
print((data.min()), data.max())
```

```python
# Boolean Indexing (critical for ML)
arr = np.array([1, -2, 3, -4, 5])
positive = arr[arr > 0] 
print(positive)
```

```python
# Practice
dt = np.random.randn(5, 5)          # random 5x5 matrix
print(dt)
print('---------------------')
print(dt.diagonal())                # print diagonal elements
print('---------------------')
row_means = np.mean(dt, axis = 1)   # compute mean of each row
print(row_means)
```
