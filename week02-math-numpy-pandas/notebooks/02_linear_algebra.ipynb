```python
import numpy as np

# Dot product (foundation of neural nets)
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
dot = np.dot(a, b)
# OR
dot = a @ b         # same thing

print(dot)
```

```python
# Matrix Multiplication
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
C = A @ B
print(C)
```

```python
# This is a nueral network layer
# y = ax + b
a = np.random.randn(3, 2)       # weights: 3 inputs → 2 outputs
x = np.random.randn(2)          # input vector
b = np.random.randn(3)          # bias vector
y = a @ x + b                   # forward pass!
print(f"Input: {x.shape}, Output: {y.shape}")
print(f"Input: {x}, Output: {y}")
```

```python
# Batch processing (multiple samples at once)
X = np.random.randn(100, 2)     # 100 samples, 2 features each
Y = X @ a.T + b                 # (100, 2) @ (2, 3).T + (3,) = (100, 3)
print(Y.shape)                  # (100, 3) — 100 predictions
```

```python
# Transpose
A = np.array(([1, 2, 3], [4, 5, 6]))
print(A.shape)
print(A.T.shape)
```

```python
# Norms (used in regularization)
v = np.array([[3, 4]])
l2_norm = np.linalg.norm(v)      # Euclidean
l1_norm = np.sum(np.abs(v))

print(f"L2 norm: {l2_norm}, L1 norm: {l1_norm}")
```

```python
# Solve linear system: Ax = b
A = np.array([[3, 1], [1, 2]])
b = np.array([9, 8])
x = np.linalg.solve(A, b)
print(x)
```

```python
# Eigenvalues (PCA uses this - week 8 preview)
matrix = np.array([[4, 2], [1, 3]])
eigenvalues, eigenvectors = np.linalg.eig(matrix)
print("Eigenvalues:", eigenvalues)
print("Eigenvectors:", eigenvectors)
```
