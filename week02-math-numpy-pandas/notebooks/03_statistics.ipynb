```python
import numpy as np
import matplotlib.pyplot as plt


#   Descriptive statistics
data = np.random.randn(1000)

print(f"Mean: {data.mean():.3f}")
print(f"Median: {np.median(data):.3f}")
print(f"Std: {data.std():.3f}")
print(f"Min: {data.min():.3f}")
print(f"Max: {data.max():.3f}")
print(f"25th Percentile: {np.percentile(data, 25):.3f}")
print(f"75th Percentile: {np.percentile(data, 75):.3f}")
```

```python
# Correlation (feature selection)
np.random.seed(42)
x = np.random.randn(100)
y = 2 * x + np.random.randn(100) * 0.5
z = np.random.randn(100)

print(f"corr(x, y): {np.corrcoef(x, y)[0, 1]:.3f}")
print(f"corr(x, z): {np.corrcoef(x, z)[0, 1]:.3f}")
```

```python
# Distributions
normal = np.random.normal(loc=0, scale=1, size=1000)
uniform = np.random.uniform(0, 1, size=1000)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].hist(normal, bins=30, edgecolor='black')
axes[0].set_title('Normal Distribution')
axes[1].hist(uniform, bins=30, edgecolor='black')
axes[1].set_title('Uniform Distribution')
plt.savefig('/Users/bharathkbk.12/ai-ml-journey/week02-math-numpy-pandas/data/distributions.png')
plt.show()
```

```python
# Bayes Theorem (classification intuition)
# P(spam | word) = P(word | spam) * P(spam) / P(word)

p_spam = 0.3
p_word_given_spam = 0.8
p_word = 0.5
p_spam_given_word = (p_word_given_spam * p_spam) / p_word
print(f"P(spam | word) = {p_spam_given_word:.3f}")
```

```python
# Implement from scratch
def mean(arr): return sum(arr) / len(arr)
def variance(arr): return sum((x - mean(arr))**2 for x in arr) / (len(arr) - 1)
def std(arr): return variance(arr) ** 0.5
```

```python
test = [1,2 ,3 , 4, 5]
assert abs(mean(test) - np.mean(test)) < 1e-10
assert abs(std(test) - np.std(test, ddof=1)) < 1e-10
print("Manual stats match numpy")
```
